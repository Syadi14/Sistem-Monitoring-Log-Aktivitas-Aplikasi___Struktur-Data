```python
# 1.4.3 (DAYA CLEANING)
```

```python
df.isnull().sum()
```

```python
# Numerik
df["rainfall_mm"] = df["rainfall_mm"].fillna(df["rainfall_mm"].median())
df["temperature_c"] = df["temperature_c"].fillna(df["temperature_c"].median())
df["fertilizer_kg_ha"] = df["fertilizer_kg_ha"].fillna(df["fertilizer_kg_ha"].median())
```

```python
# Kategorik
df["season"] = df["season"].fillna(df["season"].mode()[0])
df["region"] = df["region"].fillna(df["region"].mode()[0])
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
# 1.4.4 (Future Engineering)
```

```python
df.shape
```

```python
df.head()
```

```python
# Rainfall-sequared
df["rainfall_squared"] = df["rainfall_mm"] ** 2
```

```python
#Rainfall-deviation
df["rainfall_deviation"] = abs(df["rainfall_mm"] - df["rainfall_mm"].mean())
```

```python
#Fertilizer-efficiency
df["fertilizer_efficiency"] = df["fertilizer_kg_ha"] / df["rainfall_mm"]
```

```python
df.shape
```

```python
df.head()
```

```python
df = pd.get_dummies(df, columns=["season", "region"], drop_first=True)
```

```python
df.head()
```

```python
# 1.4.5 (SUPERVISED LEARNING)
```

```python
X = df.drop("yield_ton_ha", axis=1)
y = df["yield_ton_ha"]
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
X_numeric = X.select_dtypes(include=[np.number])
model.fit(X_numeric, y)

y_pred = model.predict(X_numeric)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

mae = mean_absolute_error(y, y_pred)
rmse = np.sqrt(mean_squared_error(y, y_pred))

print("MAE:", mae)
print("RMSE:", rmse)
```

```python
import matplotlib.pyplot as plt

plt.scatter(y, y_pred)
plt.xlabel("Actual")
plt.ylabel("Predicted")
plt.title("Actual vs Predicted")
plt.show()
```

```python
X_cluster = df[["rainfall_mm", "fertilizer_kg_ha"]]
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []
K = range(1, 6)

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_cluster)
    inertia.append(kmeans.inertia_)

plt.plot(K, inertia, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
df["cluster"] = kmeans.fit_predict(X_cluster)
```

```python
plt.scatter(df["rainfall_mm"], df["fertilizer_kg_ha"], c=df["cluster"])
plt.xlabel("Rainfall")
plt.ylabel("Fertilizer")
plt.title("Cluster Visualization")
plt.show()
```
