```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
# ========================
# TAHUN
# ========================
tahun = list(range(2035, 2056))

# ========================
# NILAI AWAL
# ========================
awal = 300  # korban awal (juta jiwa)

# ========================
# PERHITUNGAN DATA
# ========================

# Naik 1% per tahun
korban_naik = []
nilai = awal
for t in tahun:
    korban_naik.append(nilai)
    nilai = nilai * 1.01

# Turun 0.5% per tahun
korban_turun = []
nilai = awal
for t in tahun:
    korban_turun.append(nilai)
    nilai = nilai * 0.995

# ========================
# VISUALISASI
# ========================
plt.figure(figsize=(10,6))

plt.plot(tahun, korban_naik, marker='o', label='Naik 1% per tahun')
plt.plot(tahun, korban_turun, marker='o', label='Turun 0.5% per tahun')

plt.title("Perbandingan Skenario Korban Jiwa")
plt.xlabel("Tahun")
plt.ylabel("Korban (juta jiwa)")

plt.legend()
plt.grid()

plt.show()
```
