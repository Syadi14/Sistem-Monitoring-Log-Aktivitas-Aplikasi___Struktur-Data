```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# ========================
# TAHUN
# ========================
tahun = list(range(2035, 2056))

# ========================
# NILAI AWAL
# ========================
awal = 300  # korban awal (juta jiwa)

# ========================
# PERHITUNGAN DATA
# ========================

# Naik 1% per tahun
korban_naik = []
nilai = awal
for t in tahun:
    korban_naik.append(nilai)
    nilai = nilai * 1.01

# Turun 0.5% per tahun
korban_turun = []
nilai = awal
for t in tahun:
    korban_turun.append(nilai)
    nilai = nilai * 0.995

# ========================
# VISUALISASI
# ========================
plt.figure(figsize=(10,6))

plt.plot(tahun, korban_naik, marker='o', label='Naik 1% per tahun')
plt.plot(tahun, korban_turun, marker='o', label='Turun 0.5% per tahun')

plt.title("Perbandingan Skenario Korban Jiwa")
plt.xlabel("Tahun")
plt.ylabel("Korban (juta jiwa)")

plt.legend()
plt.grid()

plt.show()
```

```python
import matplotlib.pyplot as plt

# ========================
# TAHUN
# ========================
tahun = list(range(2035, 2061))

# ========================
# NILAI AWAL
# ========================
awal = 300

# ========================
# TANPA BIONEXA (MERAH)
# ========================
korban_naik = []
nilai = awal
for t in tahun:
    korban_naik.append(nilai)
    nilai = nilai * 1.01  # naik 1%

# ========================
# DENGAN BIONEXA (HIJAU)
# ========================
korban_bionexa = []
nilai = awal

for t in tahun:
    if t < 2045:
        # sebelum implementasi → masih naik
        korban_bionexa.append(nilai)
        nilai = nilai * 1.01
    else:
        # setelah BIONEXA → turun
        korban_bionexa.append(nilai)
        nilai = nilai * 0.97   # turunnya bisa kamu adjust

# ========================
# PLOT
# ========================
plt.figure(figsize=(10,6))

plt.plot(tahun, korban_naik, label='Tanpa BIONEXA', linestyle='-')
plt.plot(tahun, korban_bionexa, label='Dengan BIONEXA (mulai 2045)', linestyle='-')

# garis vertikal penanda 2045
plt.axvline(x=2045, linestyle='--')

plt.title("Simulasi Dampak Implementasi BIONEXA")
plt.xlabel("Tahun")
plt.ylabel("Korban (juta jiwa)")

plt.legend()
plt.grid()

plt.show()
```

```python
import matplotlib.pyplot as plt

# ========================
# TAHUN
# ========================
tahun = list(range(2035, 2056))

# ========================
# NILAI AWAL
# ========================
awal = 300

# ========================
# TANPA BIONEXA (MERAH)
# ========================
korban_naik = []
nilai = awal
for t in tahun:
    korban_naik.append(round(nilai, 1))  # dibulatkan
    nilai *= 1.01

# ========================
# DENGAN BIONEXA (HIJAU)
# ========================
korban_bionexa = []
nilai = awal

for t in tahun:
    if t < 2045:
        korban_bionexa.append(round(nilai, 1))
        nilai *= 1.01
    else:
        korban_bionexa.append(round(nilai, 1))
        nilai *= 0.98  # penurunan lebih smooth

# ========================
# PLOT
# ========================
plt.figure(figsize=(10,6))

plt.plot(tahun, korban_naik, marker='o', linewidth=2, label='Tanpa BIONEXA')
plt.plot(tahun, korban_bionexa, marker='o', linewidth=2, label='Dengan BIONEXA')

plt.title("Proyeksi Korban Jiwa dengan dan tanpa BIONEXA", fontsize=14)
plt.xlabel("Waktu (tahun)")
plt.ylabel("Korban (juta jiwa)")

plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

```python
import matplotlib.pyplot as plt

# ========================
# SET FONT (lebih formal)
# ========================
plt.rcParams['font.family'] = 'serif'

# ========================
# TAHUN (hingga 2055)
# ========================
tahun = list(range(2035, 2056))

# ========================
# NILAI AWAL
# ========================
awal = 300

# ========================
# TANPA BIONEXA
# ========================
korban_naik = []
nilai = awal
for t in tahun:
    korban_naik.append(round(nilai, 1))
    nilai *= 1.01

# ========================
# DENGAN BIONEXA
# ========================
korban_bionexa = []
nilai = awal

for t in tahun:
    if t < 2045:
        korban_bionexa.append(round(nilai, 1))
        nilai *= 1.01
    else:
        korban_bionexa.append(round(nilai, 1))
        nilai *= 0.98  # penurunan smooth

# ========================
# PLOT
# ========================
plt.figure(figsize=(10,6))

# garis utama
plt.plot(tahun, korban_naik, linewidth=2, label='Tanpa BIONEXA')
plt.plot(tahun, korban_bionexa, linewidth=2, label='Dengan BIONEXA')

# area selisih (ini bikin grafik "wow")
plt.fill_between(tahun, korban_bionexa, korban_naik, alpha=0.2)

# judul & label
plt.title("Simulasi Dampak BIONEXA terhadap Korban Jiwa (2035–2055)", fontsize=14)
plt.xlabel("Waktu (tahun)")
plt.ylabel("Korban (juta jiwa)")

# batas sumbu biar fokus
plt.ylim(280, max(korban_naik) + 10)

plt.legend()
plt.grid(alpha=0.3)

plt.show()
```
